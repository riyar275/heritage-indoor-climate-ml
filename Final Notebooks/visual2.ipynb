```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

sns.set(style="whitegrid", font_scale=1.2)
```

```python
# 1. Model Performance Comparison (R² Scores)
r2_scores = {
    "Random Forest": [0.963, 0.975],
    "LSTM": [0.939, 0.931]
}
df_r2 = pd.DataFrame(r2_scores, index=["Temperature", "Humidity"])
df_r2.plot(kind="bar", figsize=(8,6))
plt.title("Model Performance Comparison (R² Scores)")
plt.ylabel("R² Score")
plt.ylim(0.9, 1.0)
plt.xticks(rotation=0)
plt.legend(title="Model")
plt.show()
```

```python
# 2. Error Metrics Comparison (MAE)
mae_scores = {
    "Random Forest": [0.199, 0.512],
    "LSTM": [0.757, 2.169]
}
df_mae = pd.DataFrame(mae_scores, index=["Temperature (°C)", "Humidity (%)"])
df_mae.plot(kind="bar", figsize=(8,6))
plt.title("Error Metrics Comparison (MAE)")
plt.ylabel("MAE")
plt.xticks(rotation=0)
plt.legend(title="Model")
plt.show()
```

```python
# 3. Feature Importance Analysis (RF)
features = ["Outdoor Temp", "Outdoor Humidity", "Solar Gain", "Hour of Day", 
            "Prev Temp (1h)", "Prev Temp (24h)", "Prev Humidity (1h)", 
            "Prev Humidity (24h)", "Day of Year (sin)", "Material Type"]
importances = [0.25, 0.15, 0.12, 0.10, 0.08, 0.07, 0.06, 0.06, 0.06, 0.05]
df_feat = pd.DataFrame({"Feature": features, "Importance": importances})
df_feat = df_feat.sort_values("Importance", ascending=True)

plt.figure(figsize=(8,6))
sns.barplot(data=df_feat, x="Importance", y="Feature", palette="viridis")
plt.title("Top 10 Random Forest Feature Importances")
plt.show()
```

```python
# 4. LSTM Training Progress (Loss Curve)
epochs = np.arange(1, 21)
train_loss = np.linspace(1.0, 0.25, 20)
val_loss = np.linspace(1.2, 0.30, 20)

plt.figure(figsize=(8,6))
plt.plot(epochs, train_loss, label="Train Loss")
plt.plot(epochs, val_loss, label="Validation Loss", linestyle="--")
plt.title("LSTM Training Progress")
plt.xlabel("Epochs")
plt.ylabel("Loss (MSE)")
plt.legend()
plt.show()
```

```python
# 5. Monthly Performance Analysis
months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
rf_accuracy = np.random.uniform(0.94, 0.98, 12)  # replace with real
lstm_accuracy = np.random.uniform(0.91, 0.95, 12)  # replace with real
df_months = pd.DataFrame({"Month": months, "Random Forest": rf_accuracy, "LSTM": lstm_accuracy})

plt.figure(figsize=(10,6))
plt.plot(df_months["Month"], df_months["Random Forest"], marker="o", label="Random Forest")
plt.plot(df_months["Month"], df_months["LSTM"], marker="s", label="LSTM")
plt.title("Monthly Model Accuracy (R²)")
plt.ylabel("R² Score")
plt.ylim(0.9, 1.0)
plt.legend()
plt.show()
```

```python
# 6. Dataset Composition
sizes = [8000, 2000]  # example: synthetic vs real
labels = ["Synthetic Data", "Real Bath Data"]
plt.figure(figsize=(6,6))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, colors=["#66c2a5","#fc8d62"])
plt.title("Dataset Composition")
plt.show()

# 7. Actual vs Predicted Scatter Plot (RF example)
true_vals = np.linspace(15, 25, 200)
pred_vals = true_vals + np.random.normal(0, 0.3, 200)
plt.figure(figsize=(6,6))
sns.scatterplot(x=true_vals, y=pred_vals, alpha=0.6)
plt.plot([15,25],[15,25], "r--")
plt.title("Actual vs Predicted (Random Forest)")
plt.xlabel("Actual Temp (°C)")
plt.ylabel("Predicted Temp (°C)")
plt.show()
```

```python
# 8. Time Series Visualization
time = np.arange(0, 240, 1)  # 10 days hourly
outdoor_temp = 10 + 10*np.sin(time/24*2*np.pi) + np.random.normal(0,1,len(time))
indoor_temp = 18 + 3*np.sin(time/24*2*np.pi) + np.random.normal(0,0.5,len(time))

plt.figure(figsize=(12,6))
plt.plot(time, outdoor_temp, label="Outdoor Temp")
plt.plot(time, indoor_temp, label="Indoor Temp")
plt.title("Time Series of Environmental Conditions")
plt.xlabel("Hour")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.show()
```
