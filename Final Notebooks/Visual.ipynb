```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Set style
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("husl")

# 1. Model Performance Comparison (R² Scores)
def plot_model_performance_comparison():
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    
    # R² comparison
    models = ['LSTM', 'Random Forest']
    temp_r2 = [0.939, 0.963]
    humid_r2 = [0.931, 0.975]
    
    x = np.arange(len(models))
    width = 0.35
    
    ax1.bar(x - width/2, temp_r2, width, label='Temperature', color='#e74c3c', alpha=0.8)
    ax1.bar(x + width/2, humid_r2, width, label='Humidity', color='#3498db', alpha=0.8)
    
    ax1.set_xlabel('Model')
    ax1.set_ylabel('R² Score')
    ax1.set_title('Model Performance Comparison (R² Scores)')
    ax1.set_xticks(x)
    ax1.set_xticklabels(models)
    ax1.legend()
    ax1.set_ylim(0.9, 1.0)
    
    # Add value labels on bars
    for i, (temp, humid) in enumerate(zip(temp_r2, humid_r2)):
        ax1.text(i - width/2, temp + 0.002, f'{temp:.3f}', ha='center', fontweight='bold')
        ax1.text(i + width/2, humid + 0.002, f'{humid:.3f}', ha='center', fontweight='bold')
    
    # MAE comparison
    temp_mae = [0.757, 0.199]
    humid_mae = [2.169, 0.512]
    
    ax2.bar(x - width/2, temp_mae, width, label='Temperature', color='#e74c3c', alpha=0.8)
    ax2.bar(x + width/2, humid_mae, width, label='Humidity', color='#3498db', alpha=0.8)
    
    ax2.set_xlabel('Model')
    ax2.set_ylabel('Mean Absolute Error')
    ax2.set_title('Model Error Comparison (MAE)')
    ax2.set_xticks(x)
    ax2.set_xticklabels(models)
    ax2.legend()
    
    # Add value labels on bars
    for i, (temp, humid) in enumerate(zip(temp_mae, humid_mae)):
        ax2.text(i - width/2, temp + temp*0.02, f'{temp:.3f}', ha='center', fontweight='bold')
        ax2.text(i + width/2, humid + humid*0.02, f'{humid:.3f}', ha='center', fontweight='bold')
    
    plt.tight_layout()
    plt.savefig('model_performance_comparison.png', dpi=300, bbox_inches='tight')
    plt.show()

# Call the function
plot_model_performance_comparison()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# 2. Feature Importance Chart (Random Forest)
def plot_feature_importance():
    # Data from your Random Forest results
    features = [
        'Indoor_Humidity_lag1',
        'Indoor_Temp_lag1',
        'Indoor_Humidity_lag2',
        'Indoor_Temp_lag2',
        'Indoor_Humidity_lag3',
        'Indoor_Temp_lag3',
        'Outdoor_Temp',
        'Outdoor_Temp_roll3_mean',
        'Indoor_Humidity_lag6',
        'Temp_Diff'
    ]
    
    importance_avg = [0.199398, 0.155178, 0.122904, 0.099148, 0.067737, 
                      0.065178, 0.055269, 0.038474, 0.024764, 0.022506]
    
    # Create figure
    fig, ax = plt.subplots(figsize=(12, 8))
    
    # Create horizontal bar chart
    y_pos = np.arange(len(features))
    colors = plt.cm.viridis(np.linspace(0, 1, len(features)))
    
    bars = ax.barh(y_pos, importance_avg, color=colors, alpha=0.8, edgecolor='black')
    
    # Customize the chart
    ax.set_yticks(y_pos)
    ax.set_yticklabels(features)
    ax.set_xlabel('Feature Importance Score', fontsize=12)
    ax.set_title('Top 10 Most Important Features (Random Forest Model)', fontsize=14, fontweight='bold')
    ax.grid(True, alpha=0.3, axis='x')
    ax.invert_yaxis()  # Top feature at the top
    
    # Add value labels on bars
    for i, (bar, importance) in enumerate(zip(bars, importance_avg)):
        width = bar.get_width()
        ax.text(width + 0.005, bar.get_y() + bar.get_height()/2, 
                f'{importance:.3f}', ha='left', va='center', fontweight='bold')
    
    plt.tight_layout()
    plt.savefig('feature_importance_analysis.png', dpi=300, bbox_inches='tight')
    plt.show()

# Call the function
plot_feature_importance()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 3. LSTM Training Progress Chart
def plot_lstm_training_progress():
    # Training data from your LSTM output
    epochs = list(range(1, 21))
    training_loss = [0.0737, 0.0318, 0.0252, 0.0215, 0.0194, 0.0175, 0.0163, 0.0155, 
                    0.0141, 0.0133, 0.0124, 0.0118, 0.0111, 0.0106, 0.0099, 0.0099, 
                    0.0094, 0.0090, 0.0090, 0.0086]
    
    validation_loss = [0.0957, 0.0912, 0.0742, 0.0623, 0.0589, 0.0509, 0.0454, 0.0436, 
                      0.0443, 0.0390, 0.0370, 0.0367, 0.0343, 0.0305, 0.0292, 0.0293, 
                      0.0281, 0.0300, 0.0280, 0.0276]
    
    fig, ax = plt.subplots(figsize=(12, 8))
    
    # Plot training and validation loss
    ax.plot(epochs, training_loss, 'o-', color='#e74c3c', linewidth=2, 
            markersize=6, label='Training Loss', alpha=0.8)
    ax.plot(epochs, validation_loss, 'o-', color='#3498db', linewidth=2, 
            markersize=6, label='Validation Loss', alpha=0.8)
    
    # Fill areas under curves
    ax.fill_between(epochs, training_loss, alpha=0.2, color='#e74c3c')
    ax.fill_between(epochs, validation_loss, alpha=0.2, color='#3498db')
    
    # Customize the chart
    ax.set_xlabel('Epoch', fontsize=12)
    ax.set_ylabel('Loss (MSE)', fontsize=12)
    ax.set_title('LSTM Model Training Progress', fontsize=14, fontweight='bold')
    ax.legend(fontsize=11)
    ax.grid(True, alpha=0.3)
    
    # Add annotations for key points
    min_val_idx = validation_loss.index(min(validation_loss))
    ax.annotate(f'Best Validation Loss\n{min(validation_loss):.4f}', 
                xy=(epochs[min_val_idx], validation_loss[min_val_idx]), 
                xytext=(epochs[min_val_idx]+3, validation_loss[min_val_idx]+0.01),
                arrowprops=dict(arrowstyle='->', color='black', alpha=0.7),
                fontsize=10, ha='center')
    
    plt.tight_layout()
    plt.savefig('lstm_training_progress.png', dpi=300, bbox_inches='tight')
    plt.show()

# Call the function
plot_lstm_training_progress()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 4. Monthly Performance Analysis
def plot_monthly_performance():
    # Data from your Random Forest validation results
    months = ['May', 'June', 'July']
    temp_errors = [0.086, 0.192, 0.275]
    humidity_errors = [0.408, 0.512, 0.559]
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    
    # Bar chart comparison
    x = np.arange(len(months))
    width = 0.35
    
    bars1 = ax1.bar(x - width/2, temp_errors, width, label='Temperature MAE', 
                    color='#e74c3c', alpha=0.8, edgecolor='black')
    bars2 = ax1.bar(x + width/2, humidity_errors, width, label='Humidity MAE', 
                    color='#3498db', alpha=0.8, edgecolor='black')
    
    ax1.set_xlabel('Month')
    ax1.set_ylabel('Mean Absolute Error')
    ax1.set_title('Monthly Performance Analysis (Real Heritage Data)')
    ax1.set_xticks(x)
    ax1.set_xticklabels(months)
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # Add value labels on bars
    for bar in bars1:
        height = bar.get_height()
        ax1.text(bar.get_x() + bar.get_width()/2., height + 0.005,
                f'{height:.3f}', ha='center', va='bottom', fontweight='bold')
    
    for bar in bars2:
        height = bar.get_height()
        ax1.text(bar.get_x() + bar.get_width()/2., height + 0.01,
                f'{height:.3f}', ha='center', va='bottom', fontweight='bold')
    
    # Line chart showing trends
    ax2.plot(months, temp_errors, 'o-', linewidth=3, markersize=8, 
            color='#e74c3c', label='Temperature MAE')
    ax2.plot(months, humidity_errors, 's-', linewidth=3, markersize=8, 
            color='#3498db', label='Humidity MAE')
    
    ax2.set_xlabel('Month')
    ax2.set_ylabel('Mean Absolute Error')
    ax2.set_title('Error Trends Over Time')
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    
    # Add value annotations
    for i, (month, temp_err, hum_err) in enumerate(zip(months, temp_errors, humidity_errors)):
        ax2.annotate(f'{temp_err:.3f}', (i, temp_err), textcoords="offset points",
                    xytext=(0,10), ha='center', fontweight='bold')
        ax2.annotate(f'{hum_err:.3f}', (i, hum_err), textcoords="offset points",
                    xytext=(0,10), ha='center', fontweight='bold')
    
    plt.tight_layout()
    plt.savefig('monthly_performance_analysis.png', dpi=300, bbox_inches='tight')
    plt.show()

# Call the function
plot_monthly_performance()
```
